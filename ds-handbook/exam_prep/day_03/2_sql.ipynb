<details><summary><b>Hint 1</b></summary>

Signal: the measure is in one table and the label two tables away. Pattern: join first, then GROUP BY.

</details>

<details><summary><b>Hint 2</b></summary>

1. FROM InvoiceLine il. 2. JOIN Track t ON t.TrackId = il.TrackId. 3. JOIN Genre g ON g.GenreId = t.GenreId. 4. GROUP BY g.Name with SUM(il.UnitPrice * il.Quantity). 5. ORDER BY revenue DESC LIMIT 5.

<details><summary><b>Solution</b></summary>

```sql
SELECT g.Name AS genre,
       ROUND(SUM(il.UnitPrice * il.Quantity), 2) AS revenue,
       SUM(il.Quantity) AS units
FROM InvoiceLine il
JOIN Track t ON t.TrackId = il.TrackId
JOIN Genre g ON g.GenreId = t.GenreId
GROUP BY g.Name
ORDER BY revenue DESC
LIMIT 5
```

Result:

| genre | revenue | units |
|---|---|---|
| Rock | 826.65 | 835 |
| Latin | 382.14 | 386 |
| Metal | 261.36 | 264 |
| Alternative & Punk | 241.56 | 244 |
| TV Shows | 93.53 | 47 |

**Why:** Start from the table whose grain matches the measure (one row per sold line), then join the labels. Every InvoiceLine has exactly one Track and one Genre, so the join does not duplicate revenue.

**Common mistakes:** Using `Track.UnitPrice` instead of the price actually paid (`InvoiceLine.UnitPrice`). Joining Invoice too and summing `Invoice.Total`, which counts each invoice once per line (fan-out).

**Why:** One event row joins to exactly one video, so counts stay correct. Categories with no views simply do not appear in an inner join.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: count rows and count non-NULL values in the same group. Pattern: COUNT(*) vs COUNT(col).

</details>

<details><summary><b>Hint 2</b></summary>

1. GROUP BY platform. 2. COUNT(*) for all users. 3. COUNT(country) skips NULLs. 4. Subtract for unknown.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT platform,
       COUNT(*)                  AS users,
       COUNT(country)            AS known_country,
       COUNT(*) - COUNT(country) AS unknown_country
FROM users
GROUP BY platform
ORDER BY users DESC
```

Result:

| platform | users | known_country | unknown_country |
|---|---|---|---|
| android | 8 | 7 | 1 |
| web | 6 | 5 | 1 |
| ios | 6 | 6 | 0 |

**Why:** COUNT(col) ignores NULLs, so the difference with COUNT(*) is the number of NULLs. `SUM(CASE WHEN country IS NULL THEN 1 ELSE 0 END)` is the explicit alternative.

**Complexity:** One scan of users.

**Common mistakes:** `COUNT(country = NULL)` or `WHERE country = NULL`: comparisons with NULL are never true.

**Learn more:** [sql-aggregation](https://mahsa-agz.github.io/ds-handbook/#sql-aggregation)

</details>
