Example: if Chile had 3 invoices of 2, 4 and 6, the row would be `Chile, 3, 12.0, 4.0`.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: 'for each country' plus totals. Pattern: GROUP BY with COUNT, SUM, AVG.

</details>

<details><summary><b>Hint 2</b></summary>

1. FROM Invoice GROUP BY BillingCountry. 2. SELECT the country, COUNT(*), ROUND(SUM(Total), 2), ROUND(AVG(Total), 2). 3. ORDER BY revenue DESC LIMIT 5.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT BillingCountry AS country,
       COUNT(*)                 AS invoices,
       ROUND(SUM(Total), 2)     AS revenue,
       ROUND(AVG(Total), 2)     AS avg_invoice
FROM Invoice
GROUP BY BillingCountry
ORDER BY revenue DESC
LIMIT 5
```

Result:

| country | invoices | revenue | avg_invoice |
|---|---|---|---|
| USA | 91 | 523.06 | 5.75 |
| Canada | 56 | 303.96 | 5.43 |
| France | 35 | 195.1 | 5.57 |
| Brazil | 35 | 190.1 | 5.43 |
| Germany | 28 | 156.48 | 5.59 |

**Why:** GROUP BY collapses all invoices of a country into one row; aggregates summarise each group. ORDER BY can use the SELECT alias because it runs after SELECT. Rounding hides float noise (in floating point, 0.1 + 0.2 gives 0.30000000000000004).

**Complexity:** One scan of Invoice (412 rows) and a hash per country.

**Common mistakes:** Selecting a column that is neither grouped nor aggregated (PostgreSQL rejects it; SQLite silently picks a random row). Rounding before summing (small errors add up). Using the alias in WHERE (WHERE runs before SELECT).

### Q2. Big genres only

*easy, about 5 min*

Which genres have more than 100 tracks? Return `genre` (the genre name), `tracks` and `avg_minutes` (average track length in minutes, 2 decimals), most tracks first.

The genre name lives in `Genre`, so you need one simple join: `FROM Track t JOIN Genre g ON g.GenreId = t.GenreId`.

Expected: 5 rows.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: a condition on a group total ('more than 100 tracks'). Pattern: GROUP BY with HAVING.

</details>

<details><summary><b>Hint 2</b></summary>

1. Join Track to Genre. 2. GROUP BY the genre name. 3. HAVING COUNT(*) > 100 (not WHERE). 4. AVG(Milliseconds) / 60000.0 for minutes. 5. ORDER BY tracks DESC.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT g.Name AS genre,
       COUNT(*) AS tracks,
       ROUND(AVG(t.Milliseconds) / 60000.0, 2) AS avg_minutes
FROM Track t
JOIN Genre g ON g.GenreId = t.GenreId
GROUP BY g.Name
HAVING COUNT(*) > 100
ORDER BY tracks DESC
```

Result:

| genre | tracks | avg_minutes |
|---|---|---|
| Rock | 1297 | 4.73 |
| Latin | 579 | 3.88 |
| Metal | 374 | 5.16 |
| Alternative & Punk | 332 | 3.91 |
| Jazz | 130 | 4.86 |

**Why:** WHERE filters rows before grouping; HAVING filters groups after aggregation. A condition on COUNT(*) can only live in HAVING.

**Complexity:** One scan of Track plus a hash aggregate over 25 genres.

**Common mistakes:** `WHERE COUNT(*) > 100` is an error. Using the alias `tracks` inside HAVING works in SQLite and MySQL but not in PostgreSQL; repeat `COUNT(*)` to be portable.

**Learn more:** [sql-aggregation](https://mahsa-agz.github.io/ds-handbook/#sql-aggregation)

</details>

Example: a user who liked 5 videos adds 5 to `n_events` for `like` but only 1 to `n_users`.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: three different kinds of 'count'. Pattern: COUNT(*) vs COUNT(DISTINCT col) vs COUNT(col).

</details>

<details><summary><b>Hint 2</b></summary>

1. GROUP BY event_type. 2. COUNT(*) counts rows. 3. COUNT(DISTINCT user_id) counts unique users. 4. COUNT(watch_sec) skips NULLs.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT event_type,
       COUNT(*)                AS n_events,
       COUNT(DISTINCT user_id) AS n_users,
       COUNT(watch_sec)        AS n_with_watch_time
FROM events
GROUP BY event_type
ORDER BY n_events DESC
```

Result:

| event_type | n_events | n_users | n_with_watch_time |
|---|---|---|---|
| view | 155 | 17 | 155 |
| like | 48 | 17 | 0 |
| share | 23 | 11 | 0 |
| comment | 16 | 11 | 0 |

**Why:** COUNT(*) counts rows, COUNT(col) counts non-NULL values, COUNT(DISTINCT col) counts unique non-NULL values. Only views have watch time, so the last column is 0 for the other types.

**Complexity:** One scan of events.

**Common mistakes:** Using COUNT(user_id) when the question asks for unique users. Expecting COUNT(watch_sec) to count the NULL rows. Examiners love this question because the three counts differ.

<details><summary><b>Hint 1</b></summary>

Signal: filter rows (only views), then filter groups (at least 10). Pattern: WHERE plus GROUP BY plus HAVING.

</details>

<details><summary><b>Hint 2</b></summary>

1. WHERE event_type = 'view'. 2. GROUP BY user_id. 3. HAVING COUNT(*) >= 10. 4. SUM(watch_sec) / 60.0. 5. ORDER BY watch_minutes DESC.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT user_id,
       COUNT(*) AS views,
       ROUND(SUM(watch_sec) / 60.0, 1) AS watch_minutes
FROM events
WHERE event_type = 'view'
GROUP BY user_id
HAVING COUNT(*) >= 10
ORDER BY watch_minutes DESC
```

Result:

| user_id | views | watch_minutes |
|---|---|---|
| 20 | 14 | 9.6 |
| 3 | 18 | 8.7 |
| 14 | 14 | 8.5 |
| 6 | 19 | 7.3 |
| 1 | 11 | 6.2 |
| 2 | 10 | 5.4 |
| 18 | 11 | 5.1 |
| 13 | 14 | 4.5 |

**Why:** The row filter (views only) belongs in WHERE so likes are never counted; the group filter belongs in HAVING. Order of execution: FROM, WHERE, GROUP BY, HAVING, SELECT, ORDER BY.

**Complexity:** One scan of events plus a hash per user.

**Common mistakes:** Putting `event_type = 'view'` in HAVING (error or wrong meaning). Counting all event types as views. `SUM(watch_sec) / 60` is integer division.
