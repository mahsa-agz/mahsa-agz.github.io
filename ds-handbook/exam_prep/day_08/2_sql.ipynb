```sql
    FROM Invoice
    GROUP BY CustomerId
)

       COUNT(*) AS customers,

ORDER BY avg_spend DESC
```

<details><summary><b>Hint 2</b></summary>

```sql
    FROM events
    GROUP BY user_id
)

       ROUND(1.0 * likes / views, 3) AS like_rate
FROM per_user
WHERE views >= 5
ORDER BY like_rate DESC, user_id
```

<details><summary><b>Hint 2</b></summary>

1. CTE country_rev: revenue per country. 2. CTE total: SUM(revenue) FROM country_rev (one row). 3. SELECT ... FROM country_rev CROSS JOIN total. 4. 100.0 * revenue / all_revenue.

```sql
WITH country_rev AS (
    SELECT BillingCountry AS country, SUM(Total) AS revenue
    FROM Invoice
    GROUP BY BillingCountry
),

)
```

On the made-up product tables, build a creator scoreboard. For every creator (a user who uploaded at least one video) return `creator_id`, `country` (`unknown` if NULL), `videos` (uploaded), `views` (view events on their videos, 0 if none) and `watch_minutes` (1 decimal). Most views first, ties by creator_id.

Expected: 6 rows.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: stats at two grains (video, then creator) plus a label from a third table. Pattern: chained CTEs, LEFT JOIN to keep videos with no views.

</details>

<details><summary><b>Hint 2</b></summary>

1. CTE video_stats: videos LEFT JOIN events (views only, in ON), one row per video with views and watch_sec. 2. CTE creator_stats: GROUP BY creator_id from video_stats. 3. Join users for the country.

</details>

<details><summary><b>Solution</b></summary>

```sql
WITH video_stats AS (
    SELECT v.video_id, v.creator_id,
           COUNT(e.event_id) AS views,
           COALESCE(SUM(e.watch_sec), 0) AS watch_sec
    FROM videos v
    LEFT JOIN events e ON e.video_id = v.video_id AND e.event_type = 'view'
    GROUP BY v.video_id, v.creator_id
),
creator_stats AS (
    SELECT creator_id,
           COUNT(*) AS videos,
           SUM(views) AS views,
           ROUND(SUM(watch_sec) / 60.0, 1) AS watch_minutes
    FROM video_stats
    GROUP BY creator_id
)
SELECT c.creator_id, COALESCE(u.country, 'unknown') AS country,
       c.videos, c.views, c.watch_minutes
FROM creator_stats c
JOIN users u ON u.user_id = c.creator_id
ORDER BY c.views DESC, c.creator_id
```

Result:

| creator_id | country | videos | views | watch_minutes |
|---|---|---|---|---|
| 2 | KR | 3 | 41 | 21.5 |
| 12 | DE | 4 | 41 | 24.2 |
| 14 | KR | 4 | 27 | 11.4 |
| 9 | ID | 2 | 24 | 14.1 |
| 5 | KR | 1 | 16 | 4.5 |
| 3 | US | 1 | 6 | 1.8 |

**Why:** Aggregating per video first stops the fan-out: if you join events directly and count videos, a video with 30 views counts 30 times. Each CTE has one clear grain, which makes the query easy to debug step by step.

**Complexity:** O(videos + events + users).

**Common mistakes:** `COUNT(*)` of videos after joining events (counts views, not videos). Filtering views in WHERE, which drops videos without views and their creators' video counts.

**Learn more:** [sql-cte](https://mahsa-agz.github.io/ds-handbook/#sql-cte), [sql-left-join-nulls](https://mahsa-agz.github.io/ds-handbook/#sql-left-join-nulls)

</details>

<details><summary><b>Hint 1</b></summary>

Signal: compare a group count with an average of group counts. Pattern: subquery over a derived table inside HAVING.

</details>

<details><summary><b>Hint 2</b></summary>

1. Inner derived table: COUNT(*) per AlbumId. 2. AVG of those counts (scalar). 3. Outer: GROUP BY album, HAVING COUNT(*) > 2 * (scalar).

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT al.Title AS album, COUNT(*) AS tracks
FROM Album al
JOIN Track t ON t.AlbumId = al.AlbumId
GROUP BY al.AlbumId, al.Title
HAVING COUNT(*) > (SELECT AVG(n)
                   FROM (SELECT COUNT(*) AS n FROM Track GROUP BY AlbumId) AS per_album) * 2
ORDER BY tracks DESC, album
```

Result:

| album | tracks |
|---|---|
| Greatest Hits | 57 |
| Minha Historia | 34 |
| Unplugged | 30 |
| Lost, Season 3 | 26 |
| Lost, Season 1 | 25 |
| The Office, Season 3 | 25 |
| Battlestar Galactica (Classic), Season 1 | 24 |
| Lost, Season 2 | 24 |
| My Way: The Best Of Frank Sinatra [Disc 1] | 24 |
| Afrociberdelia | 23 |
| Heroes, Season 1 | 23 |
| Instant Karma: The Amnesty International Campaign to Save Darfur | 23 |

(17 rows, first 12 shown)

**Why:** The average album size is an average of counts, so it needs its own GROUP BY inside a subquery. The same thing as a CTE: `WITH per_album AS (...) SELECT ... HAVING COUNT(*) > 2 * (SELECT AVG(n) FROM per_album)`.
