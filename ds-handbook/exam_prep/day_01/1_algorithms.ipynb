## Setup

Run this cell once before the questions.

````python
import copy

def check(fn, cases, key=None):
    """Runs fn on each (args, expected) case and prints PASS or FAIL. key: normalise outputs (e.g. sorted)."""
    ok = 0
    for args, want in cases:
        args = args if isinstance(args, tuple) else (args,)
        try:
            got = fn(*copy.deepcopy(args))
        except Exception as e:
            got = "error: " + repr(e)
        good = (key(got) == key(want)) if (key and not isinstance(got, str)) else got == want
        ok += good
        shown = ", ".join(repr(a) for a in args)
        print(("PASS " if good else "FAIL ") + f"{fn.__name__}({shown}) -> {got!r}" + ("" if good else f"   expected {want!r}"))
    print(f"{ok}/{len(cases)} passed")
````

````python
def is_valid(s):
    pass

# tests
check(is_valid, [
    ("()", True),
    ("()[]{}", True),
    ("(]", False),
    ("([)]", False),
    ("{[]}", True),
    ("", True),
    ("((", False),
    ("))", False),
])
````

<details><summary><b>Hint 1</b></summary>

Signal: the most recent unclosed item must be matched first (last in, first out). Pattern: **stack**.

</details>

<details><summary><b>Hint 2</b></summary>

1. Map each closing bracket to its opening partner: `{')': '(', ...}`.
2. Walk the string. Opening bracket: push it.
3. Closing bracket: the stack must be non-empty and its top must be the partner; pop it. Otherwise return False.
4. At the end the stack must be empty.

</details>

<details><summary><b>Solution</b></summary>

```python
def is_valid(s):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for ch in s:
        if ch in pairs:                       # closing bracket
            if not stack or stack.pop() != pairs[ch]:
                return False
        else:                                 # opening bracket
            stack.append(ch)
    return not stack                          # leftovers are unclosed
```

**Why:** A closing bracket can only match the latest opening bracket that is still open. A stack keeps exactly that order: its top is always the bracket that must close next.

**Complexity:** O(n) time, O(n) space (all openings in the worst case).

**Common mistakes:** Forgetting the final `return not stack` (fails on `"(("`). Popping from an empty stack on `")"`. Only counting brackets, which accepts `"([)]"`.

**Learn more:** [algo-stack](https://mahsa-agz.github.io/ds-handbook/#algo-stack)

</details>

<details><summary><b>Hint 1</b></summary>

Signal: for each number you need to know at once whether its partner `target - x` appeared before. Pattern: **hashing** (a dict from value to index).

</details>

<details><summary><b>Hint 2</b></summary>

1. Keep `seen = {}` mapping value -> index.
2. For each index `i` and value `x`: compute `need = target - x`.
3. If `need` is in `seen`, return `[seen[need], i]`.
4. Only then store `seen[x] = i` (look first, then add, so x never pairs with itself).

</details>

<details><summary><b>Solution</b></summary>

```python
def two_sum(nums, target):
    seen = {}                                 # value -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
    return []                                 # not reached when a pair is guaranteed
```

**Why:** Brute force tries all pairs in O(n^2). The dict answers "have I seen the partner?" in O(1) on average, so one pass is enough. Checking before inserting handles duplicates like `[3, 3]` and stops a number from pairing with itself.

**Complexity:** O(n) time, O(n) space.

**Common mistakes:** Inserting before checking (returns `[0, 0]` for `[3, 2, 4], 6`). Sorting first, which loses the original indexes. Returning the values instead of the indexes.

**Learn more:** [algo-hashing](https://mahsa-agz.github.io/ds-handbook/#algo-hashing)

</details>

A sentence counts as a palindrome if, after you lowercase it and drop everything that is not a letter
or a digit, it reads the same from both ends. Return `True` or `False`. Try to use O(1) extra space (no cleaned copy

- `"0P"` -> `False` (digits count as characters)

*Source: [LeetCode 125](https://leetcode.com/problems/valid-palindrome/)*

````python
def is_palindrome(s):
    pass

# tests
check(is_palindrome, [
    ("A man, a plan, a canal: Panama", True),
    ("race a car", False),
    (" ", True),
    ("0P", False),
    ("ab_a", True),
    ("No 'x' in Nixon", True),
    ("Was it a car or a cat I saw?", True),
])
````

<details><summary><b>Hint 1</b></summary>

Signal: compare the two ends and move inward. Pattern: **two pointers** (one at each end).

</details>

<details><summary><b>Hint 2</b></summary>

1. `i = 0`, `j = len(s) - 1`.
2. While `i < j`: move `i` right while `s[i]` is not alphanumeric; move `j` left the same way.
3. Compare `s[i].lower()` with `s[j].lower()`; if different return False.
4. Step both pointers inward. Return True at the end.

</details>

<details><summary><b>Solution</b></summary>

```python
def is_palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if not s[i].isalnum():
            i += 1
        elif not s[j].isalnum():
            j -= 1
        elif s[i].lower() != s[j].lower():
            return False
        else:
            i += 1
            j -= 1
    return True
```

**Why:** A palindrome mirrors around its centre, so each character only needs to be compared with its mirror partner. Two pointers do that in one pass and skip the junk characters in place, without building a cleaned copy. (The one-liner `t == t[::-1]` on a cleaned `t` is also fine to mention: O(n) space.)

**Complexity:** O(n) time, O(1) extra space.

**Common mistakes:** Treating `_` as a letter (`isalnum` is False for it). Forgetting digits. Skipping a junk character and comparing in the same step without re-checking `i < j`.

**Learn more:** [algo-two-pointers](https://mahsa-agz.github.io/ds-handbook/#algo-two-pointers)

</details>

<details><summary><b>Hint 1</b></summary>

Signal: compare the words column by column (character position by position). Pattern: **string scanning** (arrays and strings).

</details>

<details><summary><b>Hint 2</b></summary>

1. Take the first word as the candidate.
2. For each position `i` of the candidate, check every other word: if the word is too short or its `i`-th
   character differs, return `candidate[:i]`.
3. If no column fails, the whole first word is the prefix.

</details>

<details><summary><b>Solution</b></summary>

```python
def longest_common_prefix(words):
    if not words:
        return ""
    first = words[0]
    for i, ch in enumerate(first):
        for w in words[1:]:
            if i == len(w) or w[i] != ch:
                return first[:i]
    return first
```

**Why:** The prefix ends at the first column where any word disagrees or runs out. Scanning column by column stops as early as possible. An alternative: sort the words and compare only the first and last.

**Complexity:** O(S) time where S is the total number of characters, O(1) extra space.

**Common mistakes:** Index error when a later word is shorter than the first. Returning the shortest word without checking the characters.

**Learn more:** [algo-arrays-strings](https://mahsa-agz.github.io/ds-handbook/#algo-arrays-strings)

</details>
