Example of the shape: a 12-minute track gives `minutes = 12.0`.

Signal: filter rows, sort, keep the first few. Pattern: plain SELECT with WHERE, ORDER BY and LIMIT.

<details><summary><b>Hint 2</b></summary>

1. FROM Track. 2. WHERE length over 600000 ms and MediaTypeId <> 3. 3. Compute minutes as `Milliseconds / 60000.0` (note the .0). 4. ORDER BY Milliseconds DESC, LIMIT 5.

```sql
SELECT Name,

```

<details><summary><b>Hint 1</b></summary>

Signal: a list of allowed values plus a text pattern. Pattern: IN and LIKE, with brackets around the OR.

</details>

<details><summary><b>Hint 2</b></summary>

1. `Country IN ('USA', 'Canada')`. 2. `Email LIKE '%@gmail.com' OR Email LIKE '%@yahoo.com'` inside brackets. 3. Join the two with AND. 4. ORDER BY Country, LastName.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT FirstName, LastName, Country, Email
FROM Customer
WHERE Country IN ('USA', 'Canada')
  AND (Email LIKE '%@gmail.com' OR Email LIKE '%@yahoo.com')
ORDER BY Country, LastName
```

Result:

| FirstName | LastName | Country | Email |
|---|---|---|---|
| Martha | Silk | Canada | marthasilk@gmail.com |
| François | Tremblay | Canada | ftremblay@gmail.com |
| Julia | Barnett | USA | jubarnett@gmail.com |
| John | Gordon | USA | johngordon22@yahoo.com |
| Heather | Leacock | USA | hleacock@gmail.com |
| Frank | Ralston | USA | fralston@gmail.com |
| Victor | Stevens | USA | vstevens@yahoo.com |

**Why:** AND binds tighter than OR. Without the brackets the condition becomes `(country AND gmail) OR yahoo`, which lets Yahoo users from any country through. `%` matches any run of characters.

**Complexity:** One scan of Customer.

**Common mistakes:** Missing brackets around the OR. `LIKE '%gmail%'` would also match `gmail.com.br`. In PostgreSQL LIKE is case sensitive (use ILIKE for case-insensitive); in SQLite LIKE ignores case for ASCII letters.

### Q5. Everyone outside the US

*easy, about 5 min*

Marketing will send a campaign to every user who is **not known to be** in the US. Users with an unknown country (NULL) must be included.

Return `user_id`, `country` (show `unknown` instead of NULL) and `platform`, by user_id.

Expected: 16 rows, including users 7 and 16 whose country is NULL.

````python
run("""

""")
````

<details><summary><b>Hint 1</b></summary>

Signal: 'not equal' on a column that has NULLs. Pattern: three-valued logic, IS NULL and COALESCE.

</details>

<details><summary><b>Hint 2</b></summary>

1. `country <> 'US'` alone drops NULL rows, because NULL <> 'US' is unknown. 2. Add `OR country IS NULL`. 3. In SELECT, `COALESCE(country, 'unknown')` replaces NULL for display.

</details>

<details><summary><b>Solution</b></summary>

```sql
SELECT user_id, COALESCE(country, 'unknown') AS country, platform
FROM users
WHERE country <> 'US' OR country IS NULL
ORDER BY user_id
```

Result:

| user_id | country | platform |
|---|---|---|
| 1 | BR | web |
| 2 | KR | ios |
| 4 | KR | web |
| 5 | KR | android |
| 6 | KR | ios |
| 7 | unknown | android |
| 9 | ID | ios |
| 10 | BR | android |
| 11 | BR | ios |
| 12 | DE | android |
| 13 | ID | android |
| 14 | KR | ios |

(16 rows, first 12 shown)

**Why:** SQL has three truth values. WHERE keeps only rows where the condition is TRUE, so a NULL country fails both `= 'US'` and `<> 'US'`. COALESCE returns its first non-NULL argument.

**Complexity:** One scan of users.

**Common mistakes:** `country != 'US'` alone (loses users 7 and 16). Writing `country = NULL` (always unknown, use IS NULL). Putting COALESCE in WHERE as well is fine but slower on big tables because an index on country cannot be used.

**Learn more:** [sql-basics](https://mahsa-agz.github.io/ds-handbook/#sql-basics), [sql-left-join-nulls](https://mahsa-agz.github.io/ds-handbook/#sql-left-join-nulls)
